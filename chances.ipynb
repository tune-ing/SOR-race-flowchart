## function to see how likely different race combos are

```python
import itertools
import pandas as pd

def race_combo_likelihoods(df, n):
    """
    Calculate likelihoods (percentages) of exactly n-race combinations.
    
    Parameters
    ----------
    df : pandas.DataFrame
        The dataframe with RAC* columns and RACNUM.
    n : int
        Size of the race combination (1=alone, 2=pairs, 3=triplets, etc.)
    
    Returns
    -------
    pandas.DataFrame
        DataFrame with combination labels and percentages.
    """
    
    race_cols = {
        "White": "RACWHT",
        "Black": "RACBLK",
        "Asian": "RACASIAN",
        "AmInd": "RACAMIND",
        "NHPI": "RACPACIS",
        "SOR": "RACOTHER"
    }
    
    results = {}
    
    for combo in itertools.combinations(race_cols.items(), n):
        races = [r for r, _ in combo]
        cols = [c for _, c in combo]
        label = "+".join(races)
        
        # must have exactly n races checked
        condition = (df["RACNUM"] == n)
        for col in cols:
            condition &= (df[col] == 2)
        
        results[label] = condition.mean() * 100
    
    return pd.DataFrame.from_dict(results, orient="index", columns=["Percent"]).sort_values("Percent", ascending=False)
```

```python
def race_combo_likelihoods_relative(df, n):
    """
    Calculate likelihoods (percentages) of exactly n-race combinations relative to the selected RACNUM group.
    
    Parameters
    ----------
    df : pandas.DataFrame
        The dataframe with RAC* columns and RACNUM.
    n : int
        Size of the race combination (1=alone, 2=pairs, 3=triplets, etc.)
    
    Returns
    -------
    pandas.DataFrame
        DataFrame with combination labels and percentages relative to the selected RACNUM group.
    """
    
    race_cols = {
        "White": "RACWHT",
        "Black": "RACBLK",
        "Asian": "RACASIAN",
        "AmInd": "RACAMIND",
        "NHPI": "RACPACIS",
        "SOR": "RACOTHER"
    }
    
    results = {}
    
    # Filter the dataframe for the selected RACNUM group
    selected_group = df[df["RACNUM"] == n]
    total_selected = len(selected_group)
    
    for combo in itertools.combinations(race_cols.items(), n):
        races = [r for r, _ in combo]
        cols = [c for _, c in combo]
        label = "+".join(races)
        
        # must have exactly n races checked
        condition = (df["RACNUM"] == n)
        for col in cols:
            condition &= (df[col] == 2)
        
        # Calculate percentage relative to the selected group
        results[label] = condition.sum() / total_selected * 100
    
    return pd.DataFrame.from_dict(results, orient="index", columns=["Percent"]).sort_values("Percent", ascending=False)
```
